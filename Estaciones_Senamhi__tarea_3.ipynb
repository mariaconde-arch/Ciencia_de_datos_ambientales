### Ejercicio 2: Datos climáticos multiestación

Crea un nuevo DataFrame con datos de 5 estaciones SENAMHI de diferentes regiones del Perú (https://www.senamhi.gob.pe/site/descarga-datos/).

Luego:
1. Paresear el nombres de la estaciones y leer los archivo.
1. Unir las db (Fecha, Est1, Est2, ...)
1. Calcula el promedio mensual de precipitación por región
2. Identifica el mes más lluvioso en cada estación
4. Exporta el resultado formateado a CSV

```python


import os
import pandas as pd
from os.path import split

# Ruta
ruta = '/content/drive/MyDrive/Colab Notebooks/Ciencia de datos ambientales/estaciones'

# Lista de archivos
files = os.listdir(ruta)

# Diccionario para guardar los DataFrames
estaciones = {}

# 1. Parsear el nombre de las estaciones y leer los archivos

for station in files:

    # Separar el nombre del archivo usando split
    _, archivo = split(station)

    # Quitar "Copia de " y ".txt"
    archivo = archivo.replace(".txt", "")

    # Separar código y nombre de estación
    codigo, nombre = archivo.split("_", 1)

    print(f"Código: {codigo}, Nombre: {nombre}")

    # Leer archivo
    station_df = pd.read_table(
        f'{ruta}/{station}',
        header=None,
        sep=r'\s+',
        na_values=['-999', '-99.9']
    )

    # Las 6 columnas
    station_df.columns = ['year', 'month', 'day', 'Pr', 'Tmax', 'Tmin']


    station_df['year'] = station_df['year'].astype(int)
    station_df['month'] = station_df['month'].astype(int)
    station_df['day'] = station_df['day'].astype(int)
    station_df['Pr'] = pd.to_numeric(station_df['Pr'], errors='coerce')
    station_df['Tmax'] = pd.to_numeric(station_df['Tmax'], errors='coerce')
    station_df['Tmin'] = pd.to_numeric(station_df['Tmin'], errors='coerce')

    station_df['Fecha'] = pd.to_datetime(
    station_df[['year', 'month', 'day']])
    estaciones[nombre] = station_df[['Fecha', 'Pr']]
```

```python
station_df.head(10)
```

```python
# Crear la base combinada
db_final = None

for nombre, df in estaciones.items():

    df = df.rename(columns={'Pr': nombre})

    if db_final is None:
        db_final = df
    else:
        db_final = pd.merge(
            db_final,
            df,
            on='Fecha',
            how='outer'
        )

db_final  = db_final.sort_values('Fecha')

print(db_final.head())
```

```python
# Crear año y mes
db_final['Año'] = db_final['Fecha'].dt.year
db_final['Mes'] = db_final['Fecha'].dt.month

# Nombres de las estaciones
nombres_estaciones = list(estaciones.keys())

# Promedio mensual
precipitacion_mensual = db_final.groupby(
    ['Año', 'Mes']
)[nombres_estaciones].mean()

print(precipitacion_mensual)
```

```python
# Promedio de precipitación para cada mes
promedio_por_mes = db_final.groupby(
    'Mes'
)[nombres_estaciones].mean()

# Mes con mayor precipitación
mes_mas_lluvioso = promedio_por_mes.idxmax()

# Precipitación correspondiente
precipitacion_maxima = promedio_por_mes.max()

for estacion in nombres_estaciones:
    print(
        f"{estacion}: mes {mes_mas_lluvioso[estacion]} "
        f"({precipitacion_maxima[estacion]:.2f} mm)"
    )
```

```python
db_final.to_csv(
    f'{ruta}/estaciones_combinadas.csv',
    index=False
)
```
